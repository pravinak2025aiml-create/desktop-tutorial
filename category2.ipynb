```python
class SimpleReflexAgent:
    def act(self, percept):
        if percept == "dirty":
            return "Clean"
        return "Move"

agent = SimpleReflexAgent()

print("Fully Observable")
rooms = {"A": "dirty", "B": "clean"}

for room, status in rooms.items():
    print(room, status, "->", agent.act(status))

print("Partially Observable")
print("Unknown -> Wait")
```

```python
class ModelBasedAgent:
    def __init__(self):
        self.known = {"A": "dirty", "B": "clean"}

    def act(self, room):
        if room in self.known:
            if self.known[room] == "dirty":
                return "Clean"
            return "Move"
        return "Explore"

agent = ModelBasedAgent()

print(agent.act("A"))
print(agent.act("C"))
```

```python
class GoalBasedAgent:
    def act(self, location, goal):
        if location == goal:
            return "Goal Reached"
        return "Move"

agent = GoalBasedAgent()

print("Static")
print(agent.act("A", "B"))

print("Dynamic")
goal = "C"
print(agent.act("A", goal))
```

```python
class UtilityBasedAgent:
    def discrete(self, state):
        if state == 1:
            return "Action 1"
        return "Action 2"

    def continuous(self, value):
        if value < 0.5:
            return "Slow"
        return "Fast"

agent = UtilityBasedAgent()

print(agent.discrete(1))
print(agent.continuous(0.8))
```

```python
import random

class UtilityBasedAgent:
    def deterministic(self):
        return "Move"

    def stochastic(self):
        return random.choice(["Move", "Wait", "Turn"])

agent = UtilityBasedAgent()

print("Deterministic:", agent.deterministic())
print("Stochastic:", agent.stochastic())
```

```python
class ModelBasedAgent:
    def episodic(self, percept):
        if percept == "dirty":
            return "Clean"
        return "Move"

    def sequential(self, states):
        actions = []
        for state in states:
            if state == "dirty":
                actions.append("Clean")
            else:
                actions.append("Move")
        return actions

agent = ModelBasedAgent()

print(agent.episodic("dirty"))
print(agent.sequential(["dirty", "clean", "dirty"]))
```
