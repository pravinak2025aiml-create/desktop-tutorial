```python
class SimpleReflexAgent:
    def act(self, percept):
        if percept == "dirty":
            return "Clean"
        return "Move"

agent = SimpleReflexAgent()

for room in ["dirty", "clean", "dirty"]:
    print("Percept:", room)
    print("Action:", agent.act(room))
```

```python
class ModelBasedAgent:
    def __init__(self):
        self.state = {"A": "dirty", "B": "dirty"}

    def update(self, room, status):
        self.state[room] = status

    def act(self):
        for room in self.state:
            if self.state[room] == "dirty":
                self.state[room] = "clean"
                return f"Clean {room}"
        return "No Action"

agent = ModelBasedAgent()

agent.update("A", "dirty")
agent.update("B", "clean")

print(agent.act())
print(agent.act())
```

```python
class GoalBasedAgent:
    def __init__(self):
        self.goal = "Home"

    def act(self, location):
        if location == self.goal:
            return "Goal Reached"
        return f"Move from {location} to {self.goal}"

agent = GoalBasedAgent()

print(agent.act("School"))
print(agent.act("Home"))
```

```python
class UtilityBasedAgent:
    def __init__(self):
        self.utilities = {
            "Route1": 60,
            "Route2": 90,
            "Route3": 75
        }

    def choose(self):
        return max(self.utilities, key=self.utilities.get)

agent = UtilityBasedAgent()

print("Best Choice:", agent.choose())
print("Utility:", agent.utilities[agent.choose()])
```
