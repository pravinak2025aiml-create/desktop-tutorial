Greedy BFS Search in Tic-tac-toe Problem

```python
def heuristic(board):
    score = 0
    for row in board:
        if row.count("O") == 2 and row.count(" ") == 1:
            score += 10
        if row.count("X") == 2 and row.count(" ") == 1:
            score -= 10
    return score
def greedy_bfs(board):
    best_score = -999
    best_move = None
    for i in range(3):
        for j in range(3):
            if board[i][j] == " ":
                board[i][j] = "O"
                score = heuristic(board)
                if score > best_score:
                    best_score = score
                    best_move = (i, j)
                board[i][j] = " "
    board[best_move[0]][best_move[1]] = "O"
board = [
    ["X", "O", " "],
    ["X", " ", " "],
    [" ", " ", "O"]
]
print("Before:\n")
for row in board:
    print(row)
greedy_bfs(board)
print("After:")
for row in board:
    print(row)

```

A* Search in Missionaries and Cannibals problem

```python
def heuristic(state):
    m, c, b = state
    return m + c
def valid(m, c):
    if m < 0 or c < 0 or m > 3 or c > 3:
        return False
    if m > 0 and m < c:
        return False
    m2 = 3 - m
    c2 = 3 - c
    if m2 > 0 and m2 < c2:
        return False
    return True
def astar():
    start = (3, 3, 0)
    goal = (0, 0, 1)
    open_list = [(start, 0)]
    visited = []
    moves = [(1, 0), (2, 0), (0, 1), (0, 2), (1, 1)]
    while open_list:
        best = min(open_list,
                   key=lambda x: x[1] + heuristic(x[0]))
        state, cost = best
        open_list.remove(best)
        if state in visited:
            continue
        visited.append(state)
        print(state)
        if state == goal:
            print("Goal Reached")
            return
        m, c, b = state
        for dm, dc in moves:
            if b == 0:
                nm = m - dm
                nc = c - dc
                nb = 1
            else:
                nm = m + dm
                nc = c + dc
                nb = 0
            new_state = (nm, nc, nb)
            if valid(nm, nc) and new_state not in visited:
                open_list.append((new_state, cost + 1))
astar()
```

AO* Search in Eight Puzzle Problem

```python
def heuristic(board):
    goal = [1, 2, 3,
            4, 5, 6,
            7, 8, 0]
    count = 0
    for i in range(9):
        if board[i] != 0 and board[i] != goal[i]:
            count += 1
    return count
def ao_star(board):
    print("Current state:")
    for i in range(0, 9, 3):
        print(board[i:i+3])
    print("Heuristic:", heuristic(board))
    if board == [1, 2, 3,
                 4, 5, 6,
                 7, 8, 0]:
        print("Goal Reached")
        return
    zero = board.index(0)
    moves = []
    if zero > 2:
        moves.append(zero - 3)
    if zero < 6:
        moves.append(zero + 3)
    if zero % 3 != 0:
        moves.append(zero - 1)
    if zero % 3 != 2:
        moves.append(zero + 1)
    best = None
    best_value = 999
    for move in moves:
        new_board = board[:]
        new_board[zero], new_board[move] = new_board[move], new_board[zero]
        value = heuristic(new_board)
        if value < best_value:
            best_value = value
            best = new_board
    ao_star(best)
board = [1, 2, 3,
         4, 0, 6,
         7, 5, 8]
ao_star(board)
```

Memory Bounded Heuristic Search in Eight Puzzle Problem

```python
def heuristic(board):
    goal = [1, 2, 3, 4, 5, 6, 7, 8, 0]
    return sum(board[i] != goal[i] for i in range(9))
def search(board, depth=0):
    print(board)
    if board == [1, 2, 3, 4, 5, 6, 7, 8, 0]:
        print("Goal Reached")
        return
    zero = board.index(0)
    moves = []
    if zero >= 3:
        moves.append(zero - 3)
    if zero < 6:
        moves.append(zero + 3)
    if zero % 3 != 0:
        moves.append(zero - 1)
    if zero % 3 != 2:
        moves.append(zero + 1)
    best = None
    best_h = 999
    for move in moves:
        new_board = board[:]
        new_board[zero], new_board[move] = new_board[move], new_board[zero]
        h = heuristic(new_board)
        if h < best_h:
            best_h = h
            best = new_board
    search(best, depth + 1)
board = [1, 2, 3,
         4, 0, 6,
         7, 5, 8]
search(board)
```
