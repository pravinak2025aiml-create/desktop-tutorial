```python
import math
import random
jobs = [2, 4, 6, 8]
def cost(schedule):
    machine1 = 0
    machine2 = 0
    for i in range(len(jobs)):
        if schedule[i] == 0:
            machine1 += jobs[i]
        else:
            machine2 += jobs[i]
    return max(machine1, machine2)
current = [0, 1, 0, 1]
current_cost = cost(current)
temperature = 100
cooling = 0.9
while temperature > 1:
    new = current.copy()
    i = random.randrange(len(jobs))
    new[i] = 1 - new[i]
    new_cost = cost(new)
    if new_cost < current_cost:
        current = new
        current_cost = new_cost
    else:
        probability = math.exp((current_cost - new_cost) / temperature)
        if random.random() < probability:
            current = new
            current_cost = new_cost
    temperature = temperature * cooling
print("Best Schedule:", current)
print("Minimum Cost:", current_cost)
```
