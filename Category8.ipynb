Medical Diagnosis

```python

import networkx as nx
import matplotlib.pyplot as plt

facts = ["fever", "cough"]
rules = {
    "flu": ["fever", "cough"],
    "rest": ["flu"],
    "recovery": ["rest"]
}

G = nx.DiGraph()
G.add_edges_from([
    ("fever", "flu"), ("cough", "flu"),
    ("flu", "rest"), ("rest", "recovery")
])

for i in range(1, 4):
    print("Iteration", i)
    for result, conditions in rules.items():
        if all(c in facts for c in conditions) and result not in facts:
            facts.append(result)
            print("New fact:", result)
            break

print("Forward Chaining:", facts)

goal = "recovery"
print("Backward Chaining:")
if goal in facts:
    print(goal, "proved")
else:
    print(goal, "not proved")

nx.draw(G, with_labels=True, node_color="lightblue",
        node_size=2500, arrows=True)
plt.title("Medical Diagnosis Knowledge Graph")
plt.show()
```
