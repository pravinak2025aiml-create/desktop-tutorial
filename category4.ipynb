```python
from collections import deque

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': []
}

def bfs(graph, start, goal):
    queue = deque([[start]])
    visited = set()

    while queue:
        path = queue.popleft()
        node = path[-1]

        if node == goal:
            return path

        if node not in visited:
            visited.add(node)

            for neighbor in graph[node]:
                queue.append(path + [neighbor])

    return None

print(bfs(graph, 'A', 'G'))
```

```python
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': []
}

def dfs(graph, start, goal):
    stack = [[start]]
    visited = set()

    while stack:
        path = stack.pop()
        node = path[-1]

        if node == goal:
            return path

        if node not in visited:
            visited.add(node)

            for neighbor in reversed(graph[node]):
                stack.append(path + [neighbor])

    return None

print(dfs(graph, 'A', 'G'))
```

```python
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': []
}

def dls(node, goal, limit, path):
    if node == goal:
        return path

    if limit == 0:
        return None

    for neighbor in graph[node]:
        result = dls(neighbor, goal, limit - 1, path + [neighbor])
        if result:
            return result

    return None

print(dls('A', 'G', 3, ['A']))
```

```python
from collections import deque

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'G'],
    'F': ['C'],
    'G': ['E']
}

def bidirectional(graph, start, goal):
    if start == goal:
        return [start]

    front = deque([start])
    back = deque([goal])

    front_visited = {start}
    back_visited = {goal}

    while front and back:
        node = front.popleft()

        for neighbor in graph[node]:
            if neighbor not in front_visited:
                front_visited.add(neighbor)
                front.append(neighbor)

        node = back.popleft()

        for neighbor in graph[node]:
            if neighbor not in back_visited:
                back_visited.add(neighbor)
                back.append(neighbor)

        common = front_visited.intersection(back_visited)

        if common:
            return common.pop()

    return None

print(bidirectional(graph, 'A', 'G'))
```
