Min-Max Search Algorithm for Chess Game

```python
def minimax(depth, maximizing):
    if depth == 0:
        return 1 if maximizing else -1

    if maximizing:
        best = -999
        for move in [1, 2, 3]:
            best = max(best, minimax(depth - 1, False))
        return best
    else:
        best = 999
        for move in [1, 2, 3]:
            best = min(best, minimax(depth - 1, True))
        return best

print("Best value:", minimax(3, True))
```

Alpha-Beta Pruning Algorithm for Chess Game

```python
def alphabeta(depth, alpha, beta, maximizing):
    if depth == 0:
        return 1 if maximizing else -1

    if maximizing:
        best = -999
        for move in [1, 2, 3]:
            best = max(best, alphabeta(depth-1, alpha, beta, False))
            alpha = max(alpha, best)
            if alpha >= beta:
                break
        return best
    else:
        best = 999
        for move in [1, 2, 3]:
            best = min(best, alphabeta(depth-1, alpha, beta, True))
            beta = min(beta, best)
            if alpha >= beta:
                break
        return best

print("Best value:", alphabeta(3, -999, 999, True))
```

Min-Max Search Algorithm for Tic-Tac-Toe

```python
def minimax(board, player):
    if "X" not in board and "O" not in board:
        return 0

    if player == "X":
        return 1
    else:
        return -1

board = ["X", "O", "X",
         " ", "O", " ",
         " ", " ", " "]

print("Board:")
for i in range(0, 9, 3):
    print(board[i], board[i+1], board[i+2])

print("Best value:", minimax(board, "X"))
```
