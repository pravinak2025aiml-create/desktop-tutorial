```python
import random

values = [3, 8, 2, 15, 7, 20, 12]

print("Exploitation")
print(max(values))

print("Exploration")
for i in range(5):
    print(random.choice(values))
```

```python
def objective(x):
    return -(x - 5) ** 2 + 25

current = 0

while True:
    left = current - 1
    right = current + 1

    best = current

    if objective(left) > objective(best):
        best = left

    if objective(right) > objective(best):
        best = right

    if best == current:
        break

    current = best

print("Best State:", current)
print("Best Value:", objective(current))
```

```python
import random
import math

def objective(x):
    return -(x - 5) ** 2 + 25

current = random.randint(0, 10)
temperature = 10

while temperature > 0.1:
    next_state = current + random.choice([-1, 1])

    if 0 <= next_state <= 10:
        diff = objective(next_state) - objective(current)

        if diff > 0 or random.random() < math.exp(diff / temperature):
            current = next_state

    temperature *= 0.9

print("Best State:", current)
print("Best Value:", objective(current))
```

```python
import random

def objective(x):
    return -(x - 5) ** 2 + 25

states = [random.randint(0, 10) for _ in range(3)]

for _ in range(5):
    successors = []

    for state in states:
        if state > 0:
            successors.append(state - 1)
        successors.append(state)
        if state < 10:
            successors.append(state + 1)

    successors = sorted(successors, key=objective, reverse=True)
    states = successors[:3]

best = max(states, key=objective)

print("Beam States:", states)
print("Best State:", best)
print("Best Value:", objective(best))
```
