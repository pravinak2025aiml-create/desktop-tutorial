Map-Coloring program

```python

import matplotlib.pyplot as plt
import networkx as nx

G = nx.Graph()
G.add_edges_from([('A','B'), ('A','C'), ('B','C')])

colors = ['Red', 'Green', 'Blue']
result = {}

for i, node in enumerate(G.nodes()):
    result[node] = colors[i]
    print("Iteration", i+1, ":", node, "=", colors[i])

print("Final Coloring:", result)

nx.draw(G, with_labels=True,
        node_color=['red', 'green', 'blue'],
        node_size=2000)
plt.show()
```

Crypt-Arithmetic Puzzle

```python

import matplotlib.pyplot as plt

for i, (A, B) in enumerate([(1, 1), (1, 2), (1, 3)], 1):
    C = A + B
    print("Iteration", i, ":", A, "+", B, "=", C)

    if A != B and C != A and C != B:
        print("Valid solution")

values = [1+1, 1+2, 1+3]
plt.plot([1, 2, 3], values, marker='o')
plt.xlabel("Iteration")
plt.ylabel("A + B")
plt.title("Crypt-Arithmetic Puzzle")
plt.xticks([1, 2, 3])
plt.grid()
plt.show()
```

Local Search for CSP

```python

import matplotlib.pyplot as plt

x = 1
iterations = []
values = []

for i in range(1, 4):
    print("Iteration", i, ":", x)
    iterations.append(i)
    values.append(x)

    if x < 5:
        x = x + 2

print("Final value:", x)

plt.plot(iterations, values, marker='o')
plt.xlabel("Iteration")
plt.ylabel("Value of x")
plt.title("Local Search")
plt.grid()
plt.show()
```
